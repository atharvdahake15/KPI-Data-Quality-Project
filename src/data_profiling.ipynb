```python

import pandas as pd
from pathlib import Path

# Locate the main project folder
PROJECT_DIR = Path.cwd().parent

# Load the datasets
data_dir = PROJECT_DIR / "data"

orders_file = data_dir / "retail-orders-raw.csv"
dictionary_file = data_dir / "retail-data-dictionary.csv"

print("Orders file exists:", orders_file.exists())
print("Dictionary file exists:", dictionary_file.exists())

if not orders_file.exists() or not dictionary_file.exists():
    raise FileNotFoundError(
        "Check that both CSV files are inside the project's data folder."
    )

orders = pd.read_csv(orders_file)
dictionary = pd.read_csv(dictionary_file)

print("\nDATASET OVERVIEW")
print("Number of rows:", orders.shape[0])
print("Number of columns:", orders.shape[1])

print("\nCOLUMN NAMES")
print(orders.columns.tolist())

print("\nFIRST FIVE ROWS")
display(orders.head())

print("\nDATA DICTIONARY")
display(dictionary.head())

print("\nMISSING VALUES")
missing = pd.DataFrame({
    "Missing Count": orders.isnull().sum(),
    "Missing Percentage": (orders.isnull().mean() * 100).round(2)
})
display(missing)

print("\nDUPLICATE ROWS")
print("Duplicate rows:", orders.duplicated().sum())

print("\nDATA TYPES")
display(orders.dtypes.to_frame("Data Type"))

print("\nUNIQUE VALUES")
display(orders.nunique().to_frame("Unique Values"))

print("\nNUMERIC SUMMARY")
display(orders.describe(include="number").T)
```

```python

# DATA QUALITY CONTRACT: EXECUTABLE CHECKS

import pandas as pd
from datetime import date

df = orders.copy()

# Normalize text for validation without changing original data
segments = df["customer_segment"].astype("string").str.strip().str.title()
categories = df["category"].astype("string").str.strip().str.title()
payments = df["payment_status"].astype("string").str.strip().str.title()

# Parse dates for validation
parsed_dates = pd.to_datetime(
    df["order_date"], format="mixed", dayfirst=False, errors="coerce"
)

checks = []

def add_check(name, failed_rows, total_rows, rule):
    failed_count = int(failed_rows.sum())
    checks.append({
        "Quality Check": name,
        "Failed Rows": failed_count,
        "Total Rows": total_rows,
        "Pass Rate (%)": round(
            (total_rows - failed_count) / total_rows * 100, 2
        ) if total_rows else 100.0,
        "Rule": rule
    })

n = len(df)

# 1. Completeness
add_check(
    "Missing Order ID",
    df["order_id"].isna() | df["order_id"].astype("string").str.strip().eq(""),
    n, "Order ID must not be missing"
)

add_check(
    "Missing City",
    df["city"].isna() | df["city"].astype("string").str.strip().eq(""),
    n, "City must not be empty"
)

# 2. Uniqueness
add_check(
    "Duplicate Order ID",
    df["order_id"].duplicated(keep=False),
    n, "Order ID must be unique according to the data dictionary"
)

# 3. Date validity
add_check(
    "Invalid Date Format or Value",
    parsed_dates.isna() |
    ~df["order_date"].astype("string").str.fullmatch(r"\d{4}-\d{2}-\d{2}"),
    n, "Order date must use YYYY-MM-DD format"
)

add_check(
    "Date Outside Allowed Range",
    parsed_dates.notna() &
    ((parsed_dates.dt.date < date(2025, 1, 1)) |
     (parsed_dates.dt.date > date.today())),
    n, "Date must be between 2025-01-01 and today"
)

# 4. Allowed category values
add_check(
    "Invalid Customer Segment",
    ~segments.isin(["Student", "Fresher", "Professional"]),
    n, "Segment must be Student, Fresher, or Professional"
)

add_check(
    "Invalid Product Category",
    ~categories.isin(["Learning Kit", "Course Access", "Mentor Session"]),
    n, "Category must match the documented allowed values"
)

# 5. Numeric validity
quantity = pd.to_numeric(df["quantity"], errors="coerce")
unit_price = pd.to_numeric(df["unit_price"], errors="coerce")
discount = pd.to_numeric(df["discount_pct"], errors="coerce")

add_check(
    "Invalid Quantity",
    quantity.isna() | (quantity <= 0),
    n, "Quantity must be a positive number"
)

add_check(
    "Invalid Unit Price",
    unit_price.isna() | (unit_price <= 0),
    n, "Unit price must be positive"
)

add_check(
    "Missing or Invalid Discount",
    discount.isna() | (discount < 0) | (discount > 100),
    n, "Discount must be present and between 0 and 100"
)

# 6. Payment status
add_check(
    "Invalid Payment Status",
    ~payments.isin(["Paid", "Pending"]),
    n, "Payment status must be Paid or Pending"
)

quality_report = pd.DataFrame(checks)

print("DATA QUALITY REPORT")
display(quality_report)

print("\nTOTAL FAILED CHECKS:")
print(int((quality_report["Failed Rows"] > 0).sum()))

print("\nROWS REQUIRING INVESTIGATION:")
display(df[
    df["order_id"].duplicated(keep=False) |
    parsed_dates.isna() |
    discount.isna()
])
```

```python

# EXPORT DATA QUALITY REPORT

from pathlib import Path

PROJECT_DIR = Path.cwd().parent
deliverables_dir = PROJECT_DIR / "deliverables"
deliverables_dir.mkdir(parents=True, exist_ok=True)

# Save quality-check results
quality_report.to_csv(
    deliverables_dir / "data_quality_report.csv",
    index=False
)

# Save rows requiring investigation
problem_rows = df[
    df["order_id"].duplicated(keep=False) |
    df["city"].isna() |
    parsed_dates.isna() |
    (quantity <= 0) |
    discount.isna() |
    ~payments.isin(["Paid", "Pending"])
]

problem_rows.to_csv(
    deliverables_dir / "rows_requiring_investigation.csv",
    index=False
)

print("Reports saved successfully!")
print("Location:", deliverables_dir)
print("Files created:")
print("- data_quality_report.csv")
print("- rows_requiring_investigation.csv")
```
